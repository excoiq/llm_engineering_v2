```python
import tiktoken

encoding = tiktoken.encoding_for_model("gpt-5-mini")

tokens_eng = encoding.encode("Hello, world!")
tokens_rus = encoding.encode("Привет, мир!")
tokens_esp = encoding.encode("Hola, mundo!")
tokens_lt = encoding.encode("Labas, pasauli!")

print('English tokens count:', len(tokens_eng))
for token_id in tokens_eng:
    token_text = encoding.decode([token_id])
    print(f"{token_id} = {token_text}")

print()

print('Russian tokens count:', len(tokens_rus))
for token_id in tokens_rus:
    token_text = encoding.decode([token_id])
    print(f"{token_id} = {token_text}")

print()

print('Spanish tokens count:', len(tokens_esp))
for token_id in tokens_esp:
    token_text = encoding.decode([token_id])
    print(f"{token_id} = {token_text}")

print()

print('Lithuanian tokens count:', len(tokens_lt))
for token_id in tokens_lt:
    token_text = encoding.decode([token_id])
    print(f"{token_id} = {token_text}")

```

```python
encoding_gpt_4_1_mini = tiktoken.encoding_for_model("gpt-4.1-mini")
encoding_gpt_5_mini = tiktoken.encoding_for_model("gpt-5-mini")
encoding_gpt_5 = tiktoken.encoding_for_model("gpt-5-2025-08-07")
encoding_gpt_5_nano = tiktoken.encoding_for_model("gpt-5-nano-2025-08-07")

text = "This is a test text to check the tokenization of different models. It seems that different models have the same number of tokens for the same text."

tokens_gpt_4_1_mini = encoding_gpt_4_1_mini.encode(text)
tokens_gpt_5_mini = encoding_gpt_5_mini.encode(text)
tokens_gpt_5 = encoding_gpt_5.encode(text)
tokens_gpt_5_nano = encoding_gpt_5_nano.encode(text)


print('GPT-4.1-mini tokens count:', len(tokens_gpt_4_1_mini))
print('GPT-5-mini tokens count:', len(tokens_gpt_5_mini))
print('GPT-5 tokens count:', len(tokens_gpt_5))
print('GPT-5-nano tokens count:', len(tokens_gpt_5_nano))



```
